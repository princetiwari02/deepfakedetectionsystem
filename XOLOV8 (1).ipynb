```python
def extract_faces_from_video(frame_folder, save_folder, max_frames=None):
    os.makedirs(save_folder, exist_ok=True)
    
    frames = sorted(os.listdir(frame_folder))
    
    if max_frames:
        frames = frames[:max_frames]
    
    for frame in frames:
        img_path = os.path.join(frame_folder, frame)
        img = cv2.imread(img_path)

        if img is None:
            continue

        results = model(img, device=0)

        if len(results[0].boxes) == 0:
            continue

        # Select largest box
        boxes = results[0].boxes.xyxy.cpu().numpy()
        areas = [(x2-x1)*(y2-y1) for x1,y1,x2,y2 in boxes]
        largest_idx = areas.index(max(areas))

        x1, y1, x2, y2 = map(int, boxes[largest_idx])
        face = img[y1:y2, x1:x2]

        if face.size == 0:
            continue

        save_path = os.path.join(save_folder, frame)
        cv2.imwrite(save_path, face)
```

```python
def extract_faces_batch(frame_folder, save_folder, batch_size=32):
    os.makedirs(save_folder, exist_ok=True)

    frames = sorted(os.listdir(frame_folder))
    frame_paths = [os.path.join(frame_folder, f) for f in frames]

    for i in range(0, len(frame_paths), batch_size):
        batch_paths = frame_paths[i:i+batch_size]
        
        results = model(batch_paths, device=0)

        for img_path, result in zip(batch_paths, results):
            img = cv2.imread(img_path)
            if img is None or len(result.boxes) == 0:
                continue

            boxes = result.boxes.xyxy.cpu().numpy()
            areas = [(x2-x1)*(y2-y1) for x1,y1,x2,y2 in boxes]
            largest_idx = areas.index(max(areas))

            x1, y1, x2, y2 = map(int, boxes[largest_idx])
            face = img[y1:y2, x1:x2]

            if face.size == 0:
                continue

            frame_name = os.path.basename(img_path)
            save_path = os.path.join(save_folder, frame_name)
            cv2.imwrite(save_path, face)
```

```python
import cv2
import os
import numpy as np

def variance_of_laplacian(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return cv2.Laplacian(gray, cv2.CV_64F).var()
```

```python
sample_folder = r"C:\DeepFake_Project\faces_yolo\real"
first_video = sorted(os.listdir(sample_folder))[0]
video_path = os.path.join(sample_folder, first_video)

frames = sorted(os.listdir(video_path))

sharp_scores = []

for frame in frames[:50]:  # test first 50
    img = cv2.imread(os.path.join(video_path, frame))
    score = variance_of_laplacian(img)
    sharp_scores.append(score)

print("Min:", min(sharp_scores))
print("Max:", max(sharp_scores))
print("Mean:", np.mean(sharp_scores))
```

```python
def select_golden_frames(video_folder, keep_ratio=0.4):
    frames = sorted(os.listdir(video_folder))
    scores = []

    for frame in frames:
        img = cv2.imread(os.path.join(video_folder, frame))
        score = variance_of_laplacian(img)
        scores.append((frame, score))

    # Sort by sharpness (descending)
    scores.sort(key=lambda x: x[1], reverse=True)

    keep_count = int(len(scores) * keep_ratio)

    golden_frames = [frame for frame, _ in scores[:keep_count]]

    return golden_frames
```

```python
import shutil

def save_golden_frames(input_video_folder, output_video_folder, keep_ratio=0.4):
    os.makedirs(output_video_folder, exist_ok=True)

    golden_frames = select_golden_frames(input_video_folder, keep_ratio)

    for frame in golden_frames:
        src = os.path.join(input_video_folder, frame)
        dst = os.path.join(output_video_folder, frame)
        shutil.copy2(src, dst)
```

```python
import torch
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader
import os
from PIL import Image
from sklearn.model_selection import train_test_split
```

```python
BASE = r"C:\DeepFake_Project\faces_yolo_golden"

REAL_PATH = os.path.join(BASE, "real")
FAKE_PATH = os.path.join(BASE, "fake")

image_paths = []
labels = []

# Real images (label = 0)
for video in os.listdir(REAL_PATH):
    video_folder = os.path.join(REAL_PATH, video)
    for img in os.listdir(video_folder):
        image_paths.append(os.path.join(video_folder, img))
        labels.append(0)

# Fake images (label = 1)
for video in os.listdir(FAKE_PATH):
    video_folder = os.path.join(FAKE_PATH, video)
    for img in os.listdir(video_folder):
        image_paths.append(os.path.join(video_folder, img))
        labels.append(1)

print("Total images:", len(image_paths))
print("Real:", labels.count(0))
print("Fake:", labels.count(1))
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    image_paths,
    labels,
    test_size=0.2,
    stratify=labels,
    random_state=42
)

print("Train size:", len(X_train))
print("Validation size:", len(X_val))
```

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),          # Swin required size
    transforms.GaussianBlur(kernel_size=3),
    transforms.Grayscale(num_output_channels=3),  # Swin expects 3 channel
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5, 0.5, 0.5],
        std=[0.5, 0.5, 0.5]
    )
])
```

```python
class DeepFakeDataset(Dataset):
    def __init__(self, image_paths, labels, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = Image.open(self.image_paths[idx]).convert("RGB")
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label
```

```python
train_dataset = DeepFakeDataset(X_train, y_train, transform=transform)
val_dataset = DeepFakeDataset(X_val, y_val, transform=transform)

print("Train dataset:", len(train_dataset))
print("Validation dataset:", len(val_dataset))
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("DataLoaders ready")
```

```python
import torchvision.models as models
import torch.nn as nn
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load pretrained Swin-Tiny
model = models.swin_t(weights=models.Swin_T_Weights.IMAGENET1K_V1)

# Replace final classification layer
num_features = model.head.in_features
model.head = nn.Linear(num_features, 2)  # 2 classes: real/fake

model = model.to(device)

print("Model loaded on:", device)
```

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()

optimizer = optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

print("Loss and optimizer ready")
```

```python
from tqdm import tqdm

def train_model(model, train_loader, val_loader, criterion, optimizer, epochs=5):
    for epoch in range(epochs):
        print(f"\nEpoch {epoch+1}/{epochs}")
        
        # ---- TRAINING ----
        model.train()
        train_loss = 0
        correct = 0
        total = 0
        
        for images, labels in tqdm(train_loader):
            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            train_loss += loss.item()
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

        train_acc = 100 * correct / total
        print(f"Train Loss: {train_loss:.4f}")
        print(f"Train Accuracy: {train_acc:.2f}%")

        # ---- VALIDATION ----
        model.eval()
        val_loss = 0
        correct = 0
        total = 0

        with torch.no_grad():
            for images, labels in tqdm(val_loader):
                images = images.to(device)
                labels = labels.to(device)

                outputs = model(images)
                loss = criterion(outputs, labels)

                val_loss += loss.item()
                _, predicted = torch.max(outputs, 1)
                total += labels.size(0)
                correct += (predicted == labels).sum().item()

        val_acc = 100 * correct / total
        print(f"Validation Loss: {val_loss:.4f}")
        print(f"Validation Accuracy: {val_acc:.2f}%")
```

```python
train_model(
    model,
    train_loader,
    val_loader,
    criterion,
    optimizer,
    epochs=2
)
```

```python
import torch
import torchvision.transforms as transforms
from torch.utils.data import Dataset, DataLoader
import torchvision.models as models
import torch.nn as nn
import torch.optim as optim
import os
from PIL import Image
from sklearn.model_selection import train_test_split
```

```python
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.GaussianBlur(kernel_size=3),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3,[0.5]*3)
])
```

```python
class DeepFakeDataset(Dataset):
    def __init__(self,image_paths,labels,transform=None):
        self.image_paths=image_paths
        self.labels=labels
        self.transform=transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self,idx):
        image=Image.open(self.image_paths[idx]).convert("RGB")
        label=self.labels[idx]

        if self.transform:
            image=self.transform(image)

        return image,label
```

```python
BASE=r"C:\DeepFake_Project\faces_yolo_golden"

REAL_PATH=os.path.join(BASE,"real")
FAKE_PATH=os.path.join(BASE,"fake")

image_paths=[]
labels=[]

for video in os.listdir(REAL_PATH):
    video_folder=os.path.join(REAL_PATH,video)
    for img in os.listdir(video_folder):
        image_paths.append(os.path.join(video_folder,img))
        labels.append(0)

for video in os.listdir(FAKE_PATH):
    video_folder=os.path.join(FAKE_PATH,video)
    for img in os.listdir(video_folder):
        image_paths.append(os.path.join(video_folder,img))
        labels.append(1)

print("Total images:",len(image_paths))
```

```python
X_train,X_val,y_train,y_val=train_test_split(
    image_paths,
    labels,
    test_size=0.2,
    stratify=labels,
    random_state=42
)
```

```python
train_dataset=DeepFakeDataset(X_train,y_train,transform=transform)
val_dataset=DeepFakeDataset(X_val,y_val,transform=transform)
```

```python
train_loader=DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

val_loader=DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)
```

```python
from tqdm import tqdm

def train_model(model, train_loader, val_loader, criterion, optimizer, epochs=1):
    
    for epoch in range(epochs):

        print(f"\nEpoch {epoch+1}/{epochs}")

        # TRAINING
        model.train()
        train_loss = 0
        correct = 0
        total = 0

        for images, labels in tqdm(train_loader):

            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

            train_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)

            correct += (predicted == labels).sum().item()

        train_acc = 100 * correct / total

        print("Train Loss:", train_loss)

        print("Train Accuracy:", train_acc)


        # VALIDATION
        model.eval()

        val_loss = 0

        correct = 0

        total = 0

        with torch.no_grad():

            for images, labels in tqdm(val_loader):

                images = images.to(device)

                labels = labels.to(device)

                outputs = model(images)

                loss = criterion(outputs, labels)

                val_loss += loss.item()

                _, predicted = torch.max(outputs, 1)

                total += labels.size(0)

                correct += (predicted == labels).sum().item()

        val_acc = 100 * correct / total

        print("Validation Loss:", val_loss)

        print("Validation Accuracy:", val_acc)
```
